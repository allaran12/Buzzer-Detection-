# Phase 6B - Temporal LSTM Prototype

Notebook ini menjalankan LSTM pada urutan postingan per akun. Postingan diurutkan kronologis dan diubah menjadi fitur temporal ringan: hashtag, panjang teks, URL, reply, quote, interval antar-postingan, serta sin/cos jam aktivitas.

> Status hasil: `PROTOTYPE_ONLY`. Label multi-post saat ini baru 20 akun sehingga hasil temporal belum stabil.

```python
from pathlib import Path
import json
import sys

ROOT = next(path for path in [Path.cwd(), *Path.cwd().parents] if (path / 'PRD.md').exists())
SRC = ROOT / 'src'
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

print('Repository:', ROOT)
print('Prototype outputs:', ROOT / 'output' / 'phase6_prototypes')
```

```python
from models.temporal_lstm_prototype import run_prototype

readiness = run_prototype(
    root=ROOT,
    hidden_size=32,
    epochs=80,
    learning_rate=0.01,
    test_size=0.30,
    device='cpu',
)
print(json.dumps(readiness, indent=2, ensure_ascii=False, default=str))
```

```python
metrics = readiness['metrics']
assert readiness['status'] == 'PROTOTYPE_ONLY'
assert metrics['n'] > 0
assert Path(readiness['artifacts']['model']).exists()
assert Path(readiness['artifacts']['training_history']).exists()
assert Path(readiness['artifacts']['confusion_matrix']).exists()
print('Temporal LSTM notebook validation: OK')
print('Sequence lengths:', readiness['data']['labeled_sequence_lengths'])
```

- Akun satu-post tetap diproses, tetapi hanya memberi sequence length 1 dan tidak cukup untuk membuktikan kemampuan memahami pola urutan.
